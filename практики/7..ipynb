```python
import time
import math

def newton_system(eps, max_iter=100):
    """
    Решение системы методом Ньютона.
    Система:
        f1(x,y) = sin(y+1) - x - 1.2 = 0
        f2(x,y) = 2*y + cos(x) - 2 = 0
    Возвращает (x, y, iterations, time_used)
    """
    # Начальное приближение
    x = 0.0
    y = 0.0
    
    iter_count = 0
    start_time = time.perf_counter()
    
    for _ in range(max_iter):
        # Значения функций
        f1 = math.sin(y + 1) - x - 1.2
        f2 = 2*y + math.cos(x) - 2
        
        # Матрица Якоби
        J11 = -1.0
        J12 = math.cos(y + 1)
        J21 = -math.sin(x)
        J22 = 2.0
        
        # Определитель
        det = J11 * J22 - J12 * J21
        if abs(det) < 1e-12:
            raise RuntimeError("Определитель Якоби близок к нулю, метод не сходится")
        
        # Обратная матрица (2x2)
        invJ11 =  J22 / det
        invJ12 = -J12 / det
        invJ21 = -J21 / det
        invJ22 =  J11 / det
        
        # Поправка: -J^(-1) * F
        delta_x = -(invJ11 * f1 + invJ12 * f2)
        delta_y = -(invJ21 * f1 + invJ22 * f2)
        
        # Обновление переменных
        x_new = x + delta_x
        y_new = y + delta_y
        
        # Критерий остановки
        err = max(abs(x_new - x), abs(y_new - y))
        x, y = x_new, y_new
        iter_count += 1
        
        if err < eps:
            break
    else:
        print(f"Предупреждение: достигнут лимит итераций ({max_iter}) при eps={eps}")
    
    elapsed = time.perf_counter() - start_time
    return x, y, iter_count, elapsed

def main():
    epsilons = [0.001, 0.0001]
    results = []
    
    for eps in epsilons:
        x, y, iters, t = newton_system(eps)
        results.append((eps, x, y, iters, t))
    
    # Вывод таблицы
    print("\n" + "="*85)
    print("Решение системы нелинейных уравнений методом Ньютона")
    print("="*85)
    print(f"{'Точность':<10} {'x':<18} {'y':<18} {'Итераций':<10} {'Время (с)':<12}")
    print("-"*85)
    for eps, x, y, iters, t in results:
        print(f"{eps:<10.4f} {x:<18.10f} {y:<18.10f} {iters:<10} {t:<12.8f}")
    print("="*85)
    
    # Сравнение
    if len(results) > 1:
        print("\nСравнение результатов при разных точностях:")
        eps1, _, _, it1, t1 = results[0]
        eps2, _, _, it2, t2 = results[1]
        print(f"  При изменении точности с {eps1} на {eps2}:")
        print(f"    Количество итераций: {it1} → {it2} (увеличилось в {it2/it1:.2f} раз)")
        print(f"    Время вычислений:    {t1:.6f} с → {t2:.6f} с (увеличилось в {t2/t1:.2f} раз)")

if __name__ == "__main__":
    main()
```
