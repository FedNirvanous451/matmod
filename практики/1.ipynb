```python
import math

def solve_quadratic(a, b, c):
    """Решение квадратного уравнения ax^2 + bx + c = 0"""
    try:
        if math.isnan(a) or math.isnan(b) or math.isnan(c):
            return "Ошибка: введены некорректные числа"
    except:
        return "Ошибка при проверке чисел"
    
    if a == 0:  
        if b == 0:
            if c == 0:
                return "Бесконечно много решений (0=0)"
            else:
                return "Нет решения (противоречие: c ≠ 0 при a=0, b=0)"
        else:
            try:
                x = -c / b
                return f"Линейное уравнение, корень: x = {x}"
            except ZeroDivisionError:
                return "Ошибка: деление на ноль (b=0)"
            except:
                return "Ошибка при вычислении линейного уравнения"
    
    try:
        D = b**2 - 4*a*c
        
        if D < 0:  
            return "Нет действительных корней (D < 0)"
        elif D == 0: 
            x = -b / (2*a)
            return f"Один корень: x = {x}"
        else:  
            sqrt_D = math.sqrt(D)
            x1 = (-b + sqrt_D) / (2*a)
            x2 = (-b - sqrt_D) / (2*a)
            return f"Два корня: x1 = {x1}, x2 = {x2}"
    except ZeroDivisionError:
        return "Ошибка: деление на ноль (a=0)"
    except ValueError as e:
        if "math domain error" in str(e):
            return "Ошибка: отрицательный дискриминант под корнем (хотя должен был проверить)"
        else:
            return f"Ошибка математических вычислений: {e}"
    except OverflowError:
        return "Ошибка: переполнение (числа слишком большие)"
    except Exception as e:
        return f"Неизвестная ошибка: {e}"

def main():
    while True:
        print("=== ЗАДАНИЕ 1: Решение квадратного уравнения ===")
        print("Введите числа")
        print("(Для выхода введите 'exit' или 'q')")
        
        try:
            a_input = input("Введите a: ")
            if a_input.lower() in ['exit', 'q']:
                print("Программа завершена")
                return
            a = float(a_input)
        except ValueError:
            print("Ошибка: a должно быть числом!")
            continue
        except KeyboardInterrupt:
            print("\nПрограмма прервана")
            return

        try:
            b_input = input("Введите b: ")
            if b_input.lower() in ['exit', 'q']:
                print("Программа завершена")
                return
            b = float(b_input)
        except ValueError:
            print("Ошибка: b должно быть числом!")
            continue
        except KeyboardInterrupt:
            print("\nПрограмма прервана")
            return
    
        try:
            c_input = input("Введите c: ")
            if c_input.lower() in ['exit', 'q']:
                print("Программа завершена")
                return
            c = float(c_input)
        except ValueError:
            print("Ошибка: c должно быть числом!")
            continue
        except KeyboardInterrupt:
            print("\nПрограмма прервана")
            return
   
        try:
            result = solve_quadratic(a, b, c)
            print("Результат:", result)
        except Exception as e:
            print(f"Критическая ошибка в решении: {e}")
        
        print("\n" + "-"*30)
        cont = input("Продолжить? (Enter - да, 'exit' или 'q' - выход): ")
        if cont.lower() in ['exit', 'q']:
            print("Программа завершена")
            break

if __name__ == "__main__":
    try:
        main()
    except KeyboardInterrupt:
        print("\nПрограмма завершена пользователем")
    except Exception as e:
        print(f"Глобальная ошибка: {e}")
```

```python
import math


def is_power_of_three(n):
    if n <= 0:
        return False
    if n == 1:
        return True
    if n % 3 != 0:
        return False
    return is_power_of_three(n // 3)

def main():
    print("\n=== ЗАДАНИЕ 2: Проверка, является ли число степенью тройки ===")
    try:
        num = int(input("Введите целое число: "))
        if is_power_of_three(num):
            n = round(math.log(num, 3))
            print(f"{num} является степенью тройки (3^{n} = {num})")
        else:
            print(f"{num} НЕ является степенью тройки")
    except ValueError:
        print("Ошибка: введите целое число!")
    except KeyboardInterrupt:
        print("\nПрограмма прервана")
    except Exception as e:
        print(f"Неизвестная ошибка во втором задании: {e}")

if __name__ == "__main__":
    while True:
        try:
            main()
        except KeyboardInterrupt:
            print("\nПрограмма завершена пользователем")
        except Exception as e:
            print(f"Глобальная ошибка: {e}")


        print("\n" + "-"*30)
        cont = input("Продолжить? (Enter - да, 'exit' или 'q' - выход): ")
        if cont.lower() in ['exit', 'q']:
            print("Программа завершена")
            break
```
