```python
import math

# Константы для Земли
R_earth = 6300000      
g_earth = 10           
T_earth = 24 * 3600    
# Константы для Марса
R_mars = 3390000       
g_mars = 3.7           
T_mars = 24.6 * 3600   
# Константы для Венеры
R_venus = 6051000      
g_venus = 8.87         
T_venus = 243 * 24 * 3600  

def height_geo(R, g, T):
    """Высота геостационарного спутника над поверхностью планеты (м)"""
    r = ((T**2 * R**2 * g) / (4 * math.pi**2))**(1/3)
    return r - R

def coverage_count(R_planet, r_geo):
    """Количество спутников для покрытия поверхности"""
    OD = R_planet**2 / r_geo
    h = R_planet - OD
    if h <= 0:
        return 1
    S_seg = 2 * math.pi * R_planet * h
    S_total = 4 * math.pi * R_planet**2
    return math.ceil(S_total / S_seg)

# Задание 1: Земля
h_earth = height_geo(R_earth, g_earth, T_earth)
r_geo_earth = R_earth + h_earth
n_earth = coverage_count(R_earth, r_geo_earth)
print("=== Задание 1: Земля ===")
print(f"Высота геостационарного спутника: {h_earth / 1000} км")
print(f"Требуется спутников для покрытия: {n_earth}")
# Задание 2: Марс и Венера
h_mars = height_geo(R_mars, g_mars, T_mars)
h_venus = height_geo(R_venus, g_venus, T_venus)
print("\n=== Задание 2: Марс ===")
print(f"Высота геостационарного спутника: {h_mars / 1000} км")
print("\n=== Задание 2: Венера ===")
print(f"Высота геостационарного спутника: {h_venus / 1000} км")
```
