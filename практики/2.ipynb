```python
def mortgage_calculator(loan_amount, years, annual_rate):
    months = years * 12
    monthly_rate = annual_rate / 100 / 12


    if monthly_rate > 0:
        monthly_payment = loan_amount * (monthly_rate * (1 + monthly_rate) ** months) / ((1 + monthly_rate) ** months - 1)
    else:
        monthly_payment = loan_amount / months


    print("Период | Задолженность на начало периода | Ежемесячный платеж | Остаток по задолженности")
    print("-" * 80)

    # Расчет для каждого месяца
    remaining_debt = loan_amount
    total_payment = 0

    for month in range(1, months + 1):
        # Задолженность на начало периода
        debt_begin = remaining_debt

        # Проценты за текущий месяц
        interest = remaining_debt * monthly_rate

        # Сумма, идущая на погашение основного долга
        principal_payment = monthly_payment - interest

        # Обновление остатка
        remaining_debt -= principal_payment

        # Корректировка для последнего месяца 
        if month == months:
            monthly_payment += remaining_debt
            remaining_debt = 0

        total_payment += monthly_payment

        
        print(f"{month:6d} | {debt_begin:15.2f} | {monthly_payment:17.2f} | {remaining_debt:15.2f}")

    print("-" * 80)
    print(f"\nИтоговая информация:")
    print(f"Сумма кредита: {loan_amount:,.2f}")
    print(f"Общая выплата: {total_payment:,.2f}")
    print(f"Переплата: {total_payment - loan_amount:,.2f}")
    print(f"Срок кредита: {years} лет ({months} месяцев)")
    print(f"Годовая ставка: {annual_rate}%")


if __name__ == "__main__":
    print("Ипотечный калькулятор")
    print("-" * 40)

    try:
        loan = float(input("Введите сумму кредита: "))
        years = int(input("Введите количество лет: "))
        rate = float(input("Введите годовую процентную ставку (%): "))

        print("\n" + "=" * 40)
        mortgage_calculator(loan, years, rate)

    except ValueError:
        print("Ошибка: введите корректные числовые значения")
```
