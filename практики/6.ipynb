```python
import time
import math

def solve_system(eps, max_iter=1000):
    """
    Решение системы методом простой итерации.
    Возвращает (x, y, iterations, elapsed_time).
    """
    # Начальное приближение
    x = 0.0
    y = 0.0
    
    iter_count = 0
    start_time = time.perf_counter()
    
    for _ in range(max_iter):
        x_new = math.sin(y + 1) - 1.2
        y_new = 1 - 0.5 * math.cos(x)   # используем старое x для y_new
        
        # Проверка сходимости по максимальной разности
        err = max(abs(x_new - x), abs(y_new - y))
        x, y = x_new, y_new
        iter_count += 1
        
        if err < eps:
            break
    else:
        print(f"Предупреждение: достигнут лимит итераций ({max_iter}) при eps={eps}")
    
    elapsed_time = time.perf_counter() - start_time
    return x, y, iter_count, elapsed_time

def main():
    epsilons = [0.001, 0.0001]
    results = []
    
    for eps in epsilons:
        x, y, iters, t = solve_system(eps)
        results.append((eps, x, y, iters, t))
    
    # Вывод таблицы
    print("\n" + "="*80)
    print("Результаты решения системы нелинейных уравнений")
    print("="*80)
    print(f"{'Точность':<10} {'x':<15} {'y':<15} {'Итераций':<10} {'Время (с)':<12}")
    print("-"*80)
    for eps, x, y, iters, t in results:
        print(f"{eps:<10.4f} {x:<15.8f} {y:<15.8f} {iters:<10} {t:<12.6f}")
    print("="*80)
    
    # Сравнение
    print("\nСравнение результатов:")
    for i in range(1, len(results)):
        eps_prev, _, _, iters_prev, t_prev = results[i-1]
        eps_curr, _, _, iters_curr, t_curr = results[i]
        print(f"  При уменьшении точности с {eps_prev} до {eps_curr}:")
        print(f"    Количество итераций увеличилось в {iters_curr/iters_prev:.2f} раз(а)")
        print(f"    Время вычислений увеличилось в {t_curr/t_prev:.2f} раз(а)")

if __name__ == "__main__":
    main()
```
