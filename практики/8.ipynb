```python
import time
import math

def gradient_method(eps, max_iter=10000):
    """
    Решение системы методом градиентного спуска.
    Минимизируем F(x,y) = f1^2 + f2^2, где
        f1 = sin(y+1) - x - 1.2
        f2 = 2*y + cos(x) - 2
    Возвращает (x, y, iterations, time_used)
    """
    # Начальное приближение
    x = 0.0
    y = 0.0
    
    # Начальный шаг
    alpha = 0.1
    iter_count = 0
    start_time = time.perf_counter()
    
    for _ in range(max_iter):
        # Вычисляем значения функций
        f1 = math.sin(y + 1) - x - 1.2
        f2 = 2*y + math.cos(x) - 2
        
        # Невязка
        F = f1*f1 + f2*f2
        
        # Частные производные
        # df1/dx = -1, df1/dy = cos(y+1)
        # df2/dx = -sin(x), df2/dy = 2
        df1_dx = -1.0
        df1_dy = math.cos(y + 1)
        df2_dx = -math.sin(x)
        df2_dy = 2.0
        
        # Градиент F: dF/dx = 2*f1*df1_dx + 2*f2*df2_dx
        grad_x = 2*f1*df1_dx + 2*f2*df2_dx
        grad_y = 2*f1*df1_dy + 2*f2*df2_dy
        
        # Попытка спуска с адаптивным шагом
        for _ in range(20):  # макс. число дроблений
            x_new = x - alpha * grad_x
            y_new = y - alpha * grad_y
            
            f1_new = math.sin(y_new + 1) - x_new - 1.2
            f2_new = 2*y_new + math.cos(x_new) - 2
            F_new = f1_new*f1_new + f2_new*f2_new
            
            # Если невязка уменьшилась – принимаем шаг
            if F_new < F:
                x, y = x_new, y_new
                # Можно слегка увеличить шаг для ускорения
                alpha *= 1.05
                break
            else:
                # Уменьшаем шаг и пробуем заново
                alpha *= 0.5
        else:
            # Не удалось уменьшить невязку – выход
            break
        
        iter_count += 1
        
        # Критерий остановки по изменению переменных
        err = max(abs(grad_x * alpha), abs(grad_y * alpha))  # приближённо = изменение
        if err < eps:
            break
    else:
        print(f"Предупреждение: достигнут лимит итераций ({max_iter}) при eps={eps}")
    
    elapsed = time.perf_counter() - start_time
    return x, y, iter_count, elapsed

def main():
    epsilons = [0.001, 0.0001]
    results = []
    
    for eps in epsilons:
        x, y, iters, t = gradient_method(eps)
        results.append((eps, x, y, iters, t))
    
    # Вывод таблицы
    print("\n" + "="*85)
    print("Решение системы нелинейных уравнений градиентным методом")
    print("="*85)
    print(f"{'Точность':<10} {'x':<18} {'y':<18} {'Итераций':<10} {'Время (с)':<12}")
    print("-"*85)
    for eps, x, y, iters, t in results:
        print(f"{eps:<10.4f} {x:<18.10f} {y:<18.10f} {iters:<10} {t:<12.8f}")
    print("="*85)
    
    # Сравнение
    if len(results) > 1:
        print("\nСравнение результатов при разных точностях:")
        eps1, _, _, it1, t1 = results[0]
        eps2, _, _, it2, t2 = results[1]
        print(f"  При изменении точности с {eps1} на {eps2}:")
        print(f"    Количество итераций: {it1} → {it2} (увеличилось в {it2/it1:.2f} раз)")
        print(f"    Время вычислений:    {t1:.6f} с → {t2:.6f} с (увеличилось в {t2/t1:.2f} раз)")

if __name__ == "__main__":
    main()
```
