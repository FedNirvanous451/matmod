```python
import math

# Константы
rho = 0.5          # плотность воздуха
g = 10.0           # ускорение свободного падения
u = 3000.0         # скорость истечения газов
def rad_from_deg(deg):
    return deg * math.pi / 180.0
def ms_from_kmh(kmh):
    return kmh * 1000.0 / 3600.0
def kg_from_ton(ton):
    return ton * 1000.0

def main():
    print("Расчёт параметров реактивного самолёта")
    print("Введите три известных параметра (четвёртый оставьте пустым).")
    print("Единицы: масса в тоннах, скорость в км/ч, угол в градусах, площадь в м².")
    # Ввод данных
    M_ton = input("Масса (тонны, по умолчанию 100): ").strip()
    v_kmh = input("Скорость (км/ч, по умолчанию 900): ").strip()
    alpha_deg = input("Угол атаки (градусы, по умолчанию 3): ").strip()
    S_input = input("Площадь крыла (м², по умолчанию неизвестна): ").strip()
    # Значения по умолчанию из примера
    M_ton = float(M_ton) if M_ton else 100.0
    v_kmh = float(v_kmh) if v_kmh else 900.0
    alpha_deg = float(alpha_deg) if alpha_deg else 3.0
    S_input = float(S_input) if S_input else None
    # Перевод в СИ
    M = kg_from_ton(M_ton)
    v = ms_from_kmh(v_kmh)
    alpha = rad_from_deg(alpha_deg)
    # Определяем, какой параметр неизвестен
    unknown = None
    if S_input is None:
        unknown = 'S'
        # Вычисляем S по формуле (1)
        S = (M * g) / (2 * rho * v**2 * alpha**2)
    elif M_ton is None:
        unknown = 'M'
        M = 2 * rho * v**2 * alpha**2 * S_input / g
    elif v_kmh is None:
        unknown = 'v'
        v = math.sqrt((M * g) / (2 * rho * alpha**2 * S_input))
    elif alpha_deg is None:
        unknown = 'alpha'
        alpha = math.sqrt((M * g) / (2 * rho * v**2 * S_input))
    else:
        print("Все параметры заданы, нечего вычислять.")
        return
     # Вывод результата для неизвестного параметра
    if unknown == 'S':
        print(f"\nПлощадь крыла S = {S} м²")
    elif unknown == 'M':
        M_ton_calc = M / 1000.0
        print(f"\nМасса самолёта M = {M_ton_calc} т")
    elif unknown == 'v':
        v_kmh_calc = v * 3.6
        print(f"\nСкорость v = {v_kmh_calc} км/ч")
    elif unknown == 'alpha':
        alpha_deg_calc = alpha * 180.0 / math.pi
        print(f"\nУгол атаки a = {alpha_deg_calc}°")
    if unknown == 'S':
        S_used = S
    else:
        S_used = S_input
    mu = (2 * rho * S_used * alpha**3 * v**2) / u   # кг/с (массовый расход)
    print(f"\nМассовый расход газов μ = {mu} кг/с")
    
if __name__ == "__main__":
    main()
```
